## 52. N-Queens 2

The n-queens puzzle is the problem of placing n queens on an n x n chessboard such that no two queens attack each other.

Given an integer n, return the number of distinct solutions to the n-queens puzzle.



Example 1:


Input: n = 4
Output: 2
Explanation: There are two distinct solutions to the 4-queens puzzle as shown.
Example 2:

Input: n = 1
Output: 1


Constraints:

- 1 <= n <= 9

```python
class Solution(object):
    def totalNQueens(self, n):
        """
        :type n: int
        :rtype: int
        """
        # Attack zones track karne ke liye Sets
        cols = set()
        pos_diag = set() # (r + c)
        neg_diag = set() # (r - c)

        def backtrack(r):
            # Base Case: Agar hum N rows complete kar gaye
            # Iska matlab humne ek valid solution dhoond liya
            if r == n:
                return 1

            count = 0

            for c in range(n):
                # Check agar ye position safe hai
                if c in cols or (r + c) in pos_diag or (r - c) in neg_diag:
                    continue

                # Step 1: Place Queen (Set me add karo)
                cols.add(c)
                pos_diag.add(r + c)
                neg_diag.add(r - c)

                # Step 2: Explore (Aage ki rows se kitne solution mile?)
                # Jo bhi result aaye, use 'count' me add kar lo
                count += backtrack(r + 1)

                # Step 3: Backtrack (Remove Queen)
                cols.remove(c)
                pos_diag.remove(r + c)
                neg_diag.remove(r - c)

            return count

        return backtrack(0)
```
