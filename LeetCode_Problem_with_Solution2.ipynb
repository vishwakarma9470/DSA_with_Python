## 52. N-Queens 2

The n-queens puzzle is the problem of placing n queens on an n x n chessboard such that no two queens attack each other.

Given an integer n, return the number of distinct solutions to the n-queens puzzle.



Example 1:


Input: n = 4
Output: 2
Explanation: There are two distinct solutions to the 4-queens puzzle as shown.
Example 2:

Input: n = 1
Output: 1


Constraints:

- 1 <= n <= 9

```python
class Solution(object):
    def totalNQueens(self, n):
        """
        :type n: int
        :rtype: int
        """
        # Attack zones track karne ke liye Sets
        cols = set()
        pos_diag = set() # (r + c)
        neg_diag = set() # (r - c)

        def backtrack(r):
            # Base Case: Agar hum N rows complete kar gaye
            # Iska matlab humne ek valid solution dhoond liya
            if r == n:
                return 1

            count = 0

            for c in range(n):
                # Check agar ye position safe hai
                if c in cols or (r + c) in pos_diag or (r - c) in neg_diag:
                    continue

                # Step 1: Place Queen (Set me add karo)
                cols.add(c)
                pos_diag.add(r + c)
                neg_diag.add(r - c)

                # Step 2: Explore (Aage ki rows se kitne solution mile?)
                # Jo bhi result aaye, use 'count' me add kar lo
                count += backtrack(r + 1)

                # Step 3: Backtrack (Remove Queen)
                cols.remove(c)
                pos_diag.remove(r + c)
                neg_diag.remove(r - c)

            return count

        return backtrack(0)
```

```python
class Solution(object):
    def totalNQueens(self, n):
        self.count = 0

        # cols, ld (left diagonal), rd (right diagonal) -> Integers (Bits)
        def dfs(row, cols, ld, rd):
            if row == n:
                self.count += 1
                return

            # Available positions nikalo (jahan 0 hai wahan Queen rakh sakte hain)
            # ~(cols | ld | rd) -> Safe spots ko 1 bana deta hai
            # & ((1 << n) - 1) -> Extra bits ko trim karta hai
            available_positions = ~(cols | ld | rd) & ((1 << n) - 1)

            while available_positions:
                # Last set bit (rightmost 1) nikalo -> Ye humari position hai
                position = available_positions & -available_positions

                # Position ko available se hatao
                available_positions &= available_positions - 1

                # Next row ke liye recursion
                # Left diag shift left (<< 1), Right diag shift right (>> 1)
                dfs(row + 1, cols | position, (ld | position) << 1, (rd | position) >> 1)

        dfs(0, 0, 0, 0)
        return self.count
```

## 53. Maximum subarrays

Explanation: The subarray [4,-1,2,1] has the largest sum 6.

Example 3:

Input: nums = [5,4,-1,7,8]
