# Set

- Sets are built-in data type in python used to store collections of uniques items. they are unordered, meaning that  the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate enteries, and performing mathematical set operations like union, itersection, difference and symmetric difference.

```python
## create a set
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set = set()
print(my_set)
```

```python
type(my_set)
```

```python
my_set = set([1,2,3,4,5])
print(my_set)
```

```python
type(my_set)
```

```python
## basics sets operation
## adding and removing elements
my_set.add(7)
print(my_set)
```

```python
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(5)
print(my_set)
```

```python
my_set.discard(5)
print(my_set)
```

```python
my_set.discard(2)
print(my_set)
```

```python
### pop method
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
## clear all the elements
my_set.clear()
print(my_set)
```

```python
## set membershift test
my_set = {1,2,3,4,5}
print(1 in my_set)
print(6 in my_set)
```

```python
## mathmatical operations
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

## union
union_set = set1.union(set2)
print(union_set)
```

```python
## intersection
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

## difference
difference_set = set1.difference(set2)
print(difference_set)

# print(set1.difference(set2))
```

```python
set2.difference(set1)
```

```python
## symmetric difference
symmetric_difference_set = set1.symmetric_difference(set2)
print(symmetric_difference_set)
```

```python
## sets method
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

## isdisjoint
print(set1.isdisjoint(set2))

## is subset
print(set1.issubset(set2))

## superset
print(set1.issuperset(set2))
```

```python
lt = [1, 2,3,4,4,5,5,6,7,8]
set(lt)
```

```python
## counting unique word in text
text = 'in this dsa tutorial we are discussing about sets'
words = text.split()
print(words)

## conver list of word to set to ge uniques ords

unique_words = set(words)
print(unique_words)
```
