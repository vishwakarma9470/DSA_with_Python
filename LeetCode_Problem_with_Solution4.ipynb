## 76. Minimum Window Substring


Given two strings s and t of lengths m and n respectively, return the minimum window substring of s such that every character in t (including duplicates) is included in the window. If there is no such substring, return the empty string "".

The testcases will be generated such that the answer is unique.



Example 1:

Input: s = "ADOBECODEBANC", t = "ABC"
Output: "BANC"
Explanation: The minimum window substring "BANC" includes 'A', 'B', and 'C' from string t.
Example 2:

Input: s = "a", t = "a"
Output: "a"
Explanation: The entire string s is the minimum window.
Example 3:

Input: s = "a", t = "aa"
Output: ""
Explanation: Both 'a's from t must be included in the window.
Since the largest window of s only has one 'a', return empty string.


Constraints:

- m == s.length
- n == t.length
- 1 <= m, n <= 10^5
- s and t consist of uppercase and lowercase English letters.


Follow up: Could you find an algorithm that runs in O(m + n) time?

```python
from collections import Counter

class Solution(object):
    def minWindow(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: str
        """
        if not t or not s:
            return ""

        # Step 1: Target frequencies count kar lo
        dict_t = Counter(t)
        required = len(dict_t)  # Kitne unique characters chahiye

        # Window ke liye variables
        l, r = 0, 0
        formed = 0  # Abhi tak kitne unique chars ki requirement puri hui
        window_counts = {}

        # Result store karne ke liye (Window Length, Left, Right)
        # Init with infinity length
        ans = float("inf"), None, None

        while r < len(s):
            # --- Expand Phase ---
            char = s[r]
            window_counts[char] = window_counts.get(char, 0) + 1

            # Agar current char t me hai aur count match ho gaya
            if char in dict_t and window_counts[char] == dict_t[char]:
                formed += 1

            # --- Contract Phase ---
            # Jab tak window valid hai, left se shrink karo
            while l <= r and formed == required:
                char = s[l]

                # Minimum length check karo
                if r - l + 1 < ans[0]:
                    ans = (r - l + 1, l, r)

                # Left char remove karo
                window_counts[char] -= 1
                if char in dict_t and window_counts[char] < dict_t[char]:
                    formed -= 1

                l += 1

            # Right pointer aage badhao
            r += 1

        # Agar ans update nahi hua to empty string, warna substring
        return "" if ans[0] == float("inf") else s[ans[1] : ans[2] + 1]
```

## 77. Combinations

Given two integers n and k, return all possible combinations of k numbers chosen from the range [1, n].

You may return the answer in any order.



Example 1:

Input: n = 4, k = 2
Output: [[1,2],[1,3],[1,4],[2,3],[2,4],[3,4]]
Explanation: There are 4 choose 2 = 6 total combinations.
Note that combinations are unordered, i.e., [1,2] and [2,1] are considered to be the same combination.
Example 2:

Input: n = 1, k = 1
Output: [[1]]
Explanation: There is 1 choose 1 = 1 total combination.


Constraints:

- 1 <= n <= 20
- 1 <= k <= n



```python
class Solution(object):
    def combine(self, n, k):
        """
        :type n: int
        :type k: int
        :rtype: List[List[int]]
        """
        result = []

        def backtrack(start, current_comb):
            # Base Case: Agar length k ho gayi, result save karo
            if len(current_comb) == k:
                # Imp: list ki copy banani padegi (list(current_comb))
                result.append(list(current_comb))
                return

            # Optimization:
            # Hum loop waha tak chalayenge jaha tak enough elements bache ho.
            # Example: n=4, k=2. Agar start=4 hai, to loop chalne ka fayda nahi
            # kyunki 4 ke baad kuch bacha hi nahi aur hume 2 elements chahiye.
            # Normal loop: range(start, n + 1)
            # Optimized loop:
            needed = k - len(current_comb)
            limit = n - needed + 1

            for i in range(start, limit + 1):
                # 1. Choose
                current_comb.append(i)

                # 2. Explore (Next number se start karo)
                backtrack(i + 1, current_comb)

                # 3. Un-choose (Backtrack)
                current_comb.pop()

        backtrack(1, [])
        return result
```

```python
from itertools import combinations

class Solution(object):
    def combine(self, n, k):
        # 1 se n tak numbers generate karo
        nums = range(1, n + 1)

        # Combinations function return karta hai iterator of tuples
        # Isliye list aur map use karke format sahi karna padega
        return [list(c) for c in combinations(nums, k)]
```

## 78. subsets

Given an integer array nums of unique elements, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.



Example 1:

Input: nums = [1,2,3]
Output: [[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]
Example 2:

Input: nums = [0]
Output: [[],[0]]


Constraints:

- 1 <= nums.length <= 10
- -10 <= nums[i] <= 10
- All the numbers of nums are unique.

```python
class Solution(object):
    def subsets(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        result = []
        n = len(nums)

        def backtrack(index, current_subset):
            # Har step par jo bhi subset bana hai, use result me daal do
            # Note: current_subset[:] likhna zaroori hai taaki copy bane (reference nahi)
            result.append(current_subset[:])

            # Loop chalao aage ke elements ke liye
            for i in range(index, n):
                # 1. Include nums[i]
                current_subset.append(nums[i])

                # 2. Move to next element (Recursion)
                backtrack(i + 1, current_subset)

                # 3. Backtrack (Remove nums[i] to explore other possibilities)
                current_subset.pop()

        backtrack(0, [])
        return result
```

```python
class Solution(object):
    def subsets(self, nums):
        result = [[]]  # Start with empty subset

        for num in nums:
            # Current result me jitne bhi subsets hain,
            # un sab me 'num' add karke naye subsets banao
            new_subsets = []
            for curr in result:
                # Purane subset ki copy lo aur num add karo
                temp = list(curr)
                temp.append(num)
                new_subsets.append(temp)

            # Naye subsets ko main result me jod do
            result += new_subsets

        return result
```
