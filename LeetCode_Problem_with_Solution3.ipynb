## 62. Unique Paths


There is a robot on an m x n grid. The robot is initially located at the top-left corner (i.e., grid[0][0]). The robot tries to move to the bottom-right corner (i.e., grid[m - 1][n - 1]). The robot can only move either down or right at any point in time.

Given the two integers m and n, return the number of possible unique paths that the robot can take to reach the bottom-right corner.

The test cases are generated so that the answer will be less than or equal to 2 * 109.



Example 1:


Input: m = 3, n = 7
Output: 28
Example 2:

Input: m = 3, n = 2
Output: 3
Explanation: From the top-left corner, there are a total of 3 ways to reach the bottom-right corner:
1. Right -> Down -> Down
2. Down -> Down -> Right
3. Down -> Right -> Down


Constraints:

- 1 <= m, n <= 100

```python
class Solution(object):
    def uniquePaths(self, m, n):
        """
        :type m: int
        :type n: int
        :rtype: int
        """
        # Step 1: Ek m x n grid banao (Initialized with 1)
        # Hum 1 se isliye shuru kar rahe hain kyunki 0th row aur 0th col
        # me pahunchne ka sirf 1 hi rasta hai.
        dp = [[1] * n for _ in range(m)]

        # Step 2: Grid fill karo (Row 1 aur Col 1 se shuru karo)
        for i in range(1, m):
            for j in range(1, n):
                # Current cell = Upar wala + Left wala
                dp[i][j] = dp[i-1][j] + dp[i][j-1]

        # Step 3: Last cell return karo
        return dp[m-1][n-1]
```

```python
class Solution(object):
    def uniquePaths(self, m, n):
        # Sirf ek row maintain karo
        row = [1] * n

        # m-1 baar loop chalao (kyunki pehli row already 1 hai)
        for i in range(m - 1):
            for j in range(1, n):
                # Nayi value = Purani value (Upar) + Left value
                row[j] += row[j-1]

        return row[-1]
```

## Approach 3: Combinatorics (Maths - The "Genius" Way)
- 🧠Agar interview mein $O(1)$ space ya sabse fastest solution poochha jaye, toh yeh Maths ka sawal hai.Logic:Start se End tak pahunchne ke liye:Total Down moves chahiye: $m - 1$Total Right moves chahiye: $n - 1$Total Steps: $(m - 1) + (n - 1) = m + n - 2$Ab sawal yeh hai: "In Total Steps mein se humein (m-1) Down steps choose karne hain."Yeh Permutation & Combination ka formula hai:$$\text{Total Paths} = \binom{N}{K} = \frac{(m+n-2)!}{(m-1)! \times (n-1)!}$$Example: $m=3, n=7$.Total steps = $2 + 6 = 8$. Down moves = 2.Choose 2 downs from 8 steps: $\binom{8}{2} = \frac{8 \times 7}{2 \times 1} = 28$.

```python
import math

class Solution(object):
    def uniquePaths(self, m, n):
        # Total steps needed
        total_steps = (m - 1) + (n - 1)
        choose_steps = m - 1

        # Python ka built-in combination function
        return math.comb(total_steps, choose_steps)
```

## 63. Unique Paths 2

You are given an m x n integer array grid. There is a robot initially located at the top-left corner (i.e., grid[0][0]). The robot tries to move to the bottom-right corner (i.e., grid[m - 1][n - 1]). The robot can only move either down or right at any point in time.

An obstacle and space are marked as 1 or 0 respectively in grid. A path that the robot takes cannot include any square that is an obstacle.

Return the number of possible unique paths that the robot can take to reach the bottom-right corner.

The testcases are generated so that the answer will be less than or equal to 2 * 109.



Example 1:


Input: obstacleGrid = [[0,0,0],[0,1,0],[0,0,0]]
Output: 2
Explanation: There is one obstacle in the middle of the 3x3 grid above.
There are two ways to reach the bottom-right corner:
1. Right -> Right -> Down -> Down
2. Down -> Down -> Right -> Right
Example 2:


Input: obstacleGrid = [[0,1],[0,0]]
Output: 1


Constraints:

- m == obstacleGrid.length
- n == obstacleGrid[i].length
- 1 <= m, n <= 100
- obstacleGrid[i][j] is 0 or 1.

```python
class Solution(object):
    def uniquePathsWithObstacles(self, obstacleGrid):
        """
        :type obstacleGrid: List[List[int]]
        :rtype: int
        """
        m = len(obstacleGrid)
        n = len(obstacleGrid[0])

        # Case 1: Agar start ya end par hi obstacle ho
        if obstacleGrid[0][0] == 1 or obstacleGrid[m-1][n-1] == 1:
            return 0

        # DP table banao
        dp = [[0] * n for _ in range(m)]

        # Start point par 1 rasta hai
        dp[0][0] = 1

        # First Column fill karo
        for i in range(1, m):
            # Agar current cell 0 hai AUR upar se rasta aa raha hai (dp[i-1][0] == 1)
            if obstacleGrid[i][0] == 0 and dp[i-1][0] == 1:
                dp[i][0] = 1
            else:
                dp[i][0] = 0 # Rasta band

        # First Row fill karo
        for j in range(1, n):
            # Agar current cell 0 hai AUR left se rasta aa raha hai
            if obstacleGrid[0][j] == 0 and dp[0][j-1] == 1:
                dp[0][j] = 1
            else:
                dp[0][j] = 0

        # Baaki Grid fill karo
        for i in range(1, m):
            for j in range(1, n):
                if obstacleGrid[i][j] == 1:
                    dp[i][j] = 0 # Obstacle hai, toh 0 ways
                else:
                    dp[i][j] = dp[i-1][j] + dp[i][j-1]

        return dp[m-1][n-1]
```

```python
class Solution(object):
    def uniquePathsWithObstacles(self, obstacleGrid):
        m = len(obstacleGrid)
        n = len(obstacleGrid[0])

        # Agar start par obstacle hai
        if obstacleGrid[0][0] == 1:
            return 0

        # 1D Array (Pehli row ke liye initialize)
        dp = [0] * n
        dp[0] = 1 # Start point

        for i in range(m):
            for j in range(n):
                # Agar obstacle hai
                if obstacleGrid[i][j] == 1:
                    dp[j] = 0 # Rasta block kar do

                # Agar obstacle nahi hai aur hum first column me nahi hain
                elif j > 0:
                    # dp[j] abhi upar wali value hold kar raha hai
                    # dp[j-1] abhi left wali value hold kar raha hai
                    dp[j] += dp[j-1]

        return dp[-1]
```

## 64. Minimum path Sum


Given a m x n grid filled with non-negative numbers, find a path from top left to bottom right, which minimizes the sum of all numbers along its path.

Note: You can only move either down or right at any point in time.



Example 1:


Input: grid = [[1,3,1],[1,5,1],[4,2,1]]
Output: 7
Explanation: Because the path 1 → 3 → 1 → 1 → 1 minimizes the sum.
Example 2:

Input: grid = [[1,2,3],[4,5,6]]
Output: 12


Constraints:

- m == grid.length
- n == grid[i].length
- 1 <= m, n <= 200
- 0 <= grid[i][j] <= 200

```python
class Solution(object):
    def minPathSum(self, grid):
        """
        :type grid: List[List[int]]
        :rtype: int
        """
        rows = len(grid)
        cols = len(grid[0])

        # Step 1: Traverse the grid
        for i in range(rows):
            for j in range(cols):

                # Start Point (0,0): Kuch mat karo
                if i == 0 and j == 0:
                    continue

                # First Row: Sirf Left se aa sakte hain
                elif i == 0:
                    grid[i][j] += grid[i][j-1]

                # First Column: Sirf Upar se aa sakte hain
                elif j == 0:
                    grid[i][j] += grid[i-1][j]

                # Baaki Cells: Minimum of (Top, Left) choose karo
                else:
                    grid[i][j] += min(grid[i-1][j], grid[i][j-1])

        # Aakhri cell me accumulated min cost hogi
        return grid[rows-1][cols-1]
```

```python
class Solution(object):
    def minPathSum(self, grid):
        m = len(grid)
        n = len(grid[0])

        # Ek row maintain karo (Initialized with infinity)
        dp = [float('inf')] * n
        dp[0] = 0 # Start point ke liye dummy 0

        for i in range(m):
            for j in range(n):
                if j == 0:
                    # First column me sirf upar se aa sakte hain
                    dp[j] = dp[j] + grid[i][j]
                else:
                    # min(Upar wala, Left wala) + Current
                    # dp[j] abhi purani row ki value hai (Upar)
                    # dp[j-1] abhi update ho chuki hai (Left)
                    dp[j] = min(dp[j], dp[j-1]) + grid[i][j]

        return dp[-1]
```

## 65. Valid Number


Given a string s, return whether s is a valid number.

For example, all the following are valid numbers: "2", "0089", "-0.1", "+3.14", "4.", "-.9", "2e10", "-90E3", "3e+7", "+6e-1", "53.5e93", "-123.456e789", while the following are not valid numbers: "abc", "1a", "1e", "e3", "99e2.5", "--6", "-+3", "95a54e53".

Formally, a valid number is defined using one of the following definitions:

An integer number followed by an optional exponent.
A decimal number followed by an optional exponent.
An integer number is defined with an optional sign '-' or '+' followed by digits.

A decimal number is defined with an optional sign '-' or '+' followed by one of the following definitions:

Digits followed by a dot '.'.
Digits followed by a dot '.' followed by digits.
A dot '.' followed by digits.
An exponent is defined with an exponent notation 'e' or 'E' followed by an integer number.

The digits are defined as one or more digits.



Example 1:

Input: s = "0"

Output: true

Example 2:

Input: s = "e"

Output: false

Example 3:

Input: s = "."

Output: false



Constraints:

- 1 <= s.length <= 20
- s consists of only English letters (both uppercase and lowercase), digits (0-9), plus '+', minus '-', or dot '.'.

```python
class Solution(object):
    def isNumber(self, s):
        """
        :type s: str
        :rtype: bool
        """
        seen_digit = False
        seen_exponent = False
        seen_dot = False

        for i, char in enumerate(s):
            # Case 1: Agar digit hai
            if char.isdigit():
                seen_digit = True

            # Case 2: Agar Sign (+/-) hai
            elif char in ["+", "-"]:
                # Sign valid tabhi hai agar:
                # 1. Ye pehla character ho (i == 0)
                # 2. Ya fir iske just pehle 'e'/'E' aaya ho
                if i > 0 and s[i-1] not in ["e", "E"]:
                    return False

            # Case 3: Agar Exponent (e/E) hai
            elif char in ["e", "E"]:
                # Exponent valid tabhi hai agar:
                # 1. Pehle koi exponent nahi aaya ho
                # 2. Pehle koi digit aa chuki ho (e.g. "e3" galat hai, "1e3" sahi hai)
                if seen_exponent or not seen_digit:
                    return False

                seen_exponent = True
                # Reset digit flag kyunki 'e' ke baad hume naya number chahiye
                seen_digit = False

            # Case 4: Agar Dot (.) hai
            elif char == ".":
                # Dot valid tabhi hai agar:
                # 1. Pehle dot na aaya ho
                # 2. Pehle exponent na aaya ho (decimals exponent me allowed nahi hain)
                if seen_dot or seen_exponent:
                    return False
                seen_dot = True

            # Case 5: Koi aur character (invalid)
            else:
                return False

        # Loop ke baad, valid tabhi hai agar humne koi digit dekhi ho.
        # Ye "1e" jaise cases ko handle karta hai (seen_digit False ho jayega 'e' ke baad)
        return seen_digit
```

```python
class Solution(object):
    def isNumber(self, s):
        # States ki numbering define karte hain clarity ke liye
        # transitions[state][input_type] = next_state

        # Input Types:
        # digit: 0-9
        # sign: +, -
        # dot: .
        # exp: e, E

        state = 0 # Start State
        finals = {2, 4, 7} # Valid ending states

        # Transition Table (List of Dictionaries)
        # Har state batata hai ki agla character aane par kahan jana hai
        transitions = [
            # State 0 (Start): sign -> 1, digit -> 2, dot -> 3
            {"sign": 1, "digit": 2, "dot": 3},

            # State 1 (Sign at start): digit -> 2, dot -> 3
            {"digit": 2, "dot": 3},

            # State 2 (Integer digit): digit -> 2, dot -> 4, exp -> 5
            {"digit": 2, "dot": 4, "exp": 5},

            # State 3 (Dot without int): digit -> 4
            {"digit": 4},

            # State 4 (Decimal digit): digit -> 4, exp -> 5
            {"digit": 4, "exp": 5},

            # State 5 (Exp start): sign -> 6, digit -> 7
            {"sign": 6, "digit": 7},

            # State 6 (Exp sign): digit -> 7
            {"digit": 7},

            # State 7 (Exp digit): digit -> 7
            {"digit": 7}
        ]

        for char in s:
            input_type = ""
            if char.isdigit():
                input_type = "digit"
            elif char in "+-":
                input_type = "sign"
            elif char in "eE":
                input_type = "exp"
            elif char == ".":
                input_type = "dot"
            else:
                return False # Invalid character (e.g., 'a', space)

            # Check karo kya current state se is input ke liye koi rasta hai?
            if input_type not in transitions[state]:
                return False

            # Move to next state
            state = transitions[state][input_type]

        # Check karo kya hum valid final state par ruke hain?
        return state in finals
```

```python
import re

class Solution(object):
    def isNumber(self, s):
        # Regex pattern
        pattern = re.compile(r'^[+-]?((\d+\.?\d*)|(\.\d+))([eE][+-]?\d+)?$')

        # Match check karo
        return bool(pattern.match(s))
```

## The "Try-Except" Trick (Cheat Code) 🏴‍☠️

Hum try karenge convert karne ka. Agar error aaye, toh False.

```python
class Solution(object):
    def isNumber(self, s):
        # Edge Cases: inf aur NaN allowed nahi hain is problem me
        if "inf" in s.lower() or "nan" in s.lower():
            return False

        try:
            float(s) # Try converting to float
            return True
        except ValueError:
            return False
```

## 66. Plus One

You are given a large integer represented as an integer array digits, where each digits[i] is the ith digit of the integer. The digits are ordered from most significant to least significant in left-to-right order. The large integer does not contain any leading 0's.

Increment the large integer by one and return the resulting array of digits.



Example 1:

Input: digits = [1,2,3]
Output: [1,2,4]
Explanation: The array represents the integer 123.
Incrementing by one gives 123 + 1 = 124.
Thus, the result should be [1,2,4].
Example 2:

Input: digits = [4,3,2,1]
Output: [4,3,2,2]
Explanation: The array represents the integer 4321.
Incrementing by one gives 4321 + 1 = 4322.
Thus, the result should be [4,3,2,2].
Example 3:

Input: digits = [9]
Output: [1,0]
Explanation: The array represents the integer 9.
Incrementing by one gives 9 + 1 = 10.
Thus, the result should be [1,0].


Constraints:

- 1 <= digits.length <= 100
- 0 <= digits[i] <= 9
- digits does not contain any leading 0's.



```python
class Solution(object):
    def plusOne(self, digits):
        """
        :type digits: List[int]
        :rtype: List[int]
        """
        n = len(digits)

        # Step 1: Peeche se loop chalao (Right to Left)
        for i in range(n - 1, -1, -1):

            # Case 1: Agar digit 9 se chhota hai
            if digits[i] < 9:
                digits[i] += 1
                return digits  # Kaam khatam, yahi se wapas jao

            # Case 2: Agar digit 9 hai
            else:
                digits[i] = 0  # 9 ko 0 banao, carry next loop me handle hoga

        # Step 3: Special Case (Agar loop pura chal gaya aur return nahi hua)
        # Iska matlab saare digits 9 the (e.g., 999 -> 000)
        # Hume shuru me 1 lagana padega
        return [1] + digits
```

```python
class Solution(object):
    def plusOne(self, digits):
        # 1. Array to String -> Join -> Integer
        num = int("".join(map(str, digits)))

        # 2. Add 1
        num += 1

        # 3. Integer to String -> Map -> List
        return [int(x) for x in str(num)]
```

## 67. Add Binary

Given two binary strings a and b, return their sum as a binary string.



Example 1:

Input: a = "11", b = "1"
Output: "100"
Example 2:

Input: a = "1010", b = "1011"
Output: "10101"


Constraints:

- 1 <= a.length, b.length <= 104
- a and b consist only of '0' or '1' characters.
- Each string does not contain leading zeros except for the zero itself.

```python
class Solution(object):
    def addBinary(self, a, b):
        """
        :type a: str
        :type b: str
        :rtype: str
        """
        res = []
        i, j = len(a) - 1, len(b) - 1
        carry = 0

        # Loop tab tak chalega jab tak kisi bhi string me digits hain
        # ya fir carry bacha hua hai
        while i >= 0 or j >= 0 or carry:
            # Agar index valid hai to value lo, warna 0 maano
            digit_a = int(a[i]) if i >= 0 else 0
            digit_b = int(b[j]) if j >= 0 else 0

            total = digit_a + digit_b + carry

            # Binary result (0 ya 1)
            res.append(str(total % 2))

            # Naya Carry (0 ya 1)
            carry = total // 2

            i -= 1
            j -= 1

        # Result reverse karke string bana do
        return "".join(res[::-1])
```

```python
class Solution(object):
    def addBinary(self, a, b):
        # int(a, 2) -> Binary string 'a' ko integer me convert karo
        # bin(...) -> Sum ko wapas binary string ("0b...") me convert karo
        # [2:] -> Starting ka "0b" hata do
        return bin(int(a, 2) + int(b, 2))[2:]
```

## 68. Text Justification

Given an array of strings words and a width maxWidth, format the text such that each line has exactly maxWidth characters and is fully (left and right) justified.

You should pack your words in a greedy approach; that is, pack as many words as you can in each line. Pad extra spaces ' ' when necessary so that each line has exactly maxWidth characters.

Extra spaces between words should be distributed as evenly as possible. If the number of spaces on a line does not divide evenly between words, the empty slots on the left will be assigned more spaces than the slots on the right.

For the last line of text, it should be left-justified, and no extra space is inserted between words.

Note:

A word is defined as a character sequence consisting of non-space characters only.
Each word's length is guaranteed to be greater than 0 and not exceed maxWidth.
The input array words contains at least one word.


Example 1:

Input: words = ["This", "is", "an", "example", "of", "text", "justification."], maxWidth = 16
Output:
[
   "This    is    an",
   "example  of text",
   "justification.  "
]
Example 2:

Input: words = ["What","must","be","acknowledgment","shall","be"], maxWidth = 16
Output:
[
  "What   must   be",
  "acknowledgment  ",
  "shall be        "
]
Explanation: Note that the last line is "shall be    " instead of "shall     be", because the last line must be left-justified instead of fully-justified.
Note that the second line is also left-justified because it contains only one word.
Example 3:

Input: words = ["Science","is","what","we","understand","well","enough","to","explain","to","a","computer.","Art","is","everything","else","we","do"], maxWidth = 20
Output:
[
  "Science  is  what we",
  "understand      well",
  "enough to explain to",
  "a  computer.  Art is",
  "everything  else  we",
  "do                  "
]


Constraints:

- 1 <= words.length <= 300
- 1 <= words[i].length <= 20
- words[i] consists of only English letters and symbols.
- 1 <= maxWidth <= 100
- words[i].length <= maxWidth

```python
class Solution(object):
    def fullJustify(self, words, maxWidth):
        """
        :type words: List[str]
        :type maxWidth: int
        :rtype: List[str]
        """
        res = []
        cur_line = []
        cur_length = 0 # Sirf characters ki length (bina space ke)

        for word in words:
            # Check: Kya current words + naya word + minimum spaces (len(cur_line)) fit hoga?
            # len(cur_line) isliye add kiya kyunki har word ke baad kam se kam 1 space chahiye
            if cur_length + len(word) + len(cur_line) > maxWidth:

                # --- Line Full ho gayi, ab Justify karo ---

                # Total spaces jo hume bharne hain
                total_spaces = maxWidth - cur_length

                # Case 1: Agar line me sirf ek hi word hai
                if len(cur_line) == 1:
                    # Word + Saare spaces right side me
                    res.append(cur_line[0] + ' ' * total_spaces)

                # Case 2: Normal justification (Multiple words)
                else:
                    gaps = len(cur_line) - 1
                    space_per_gap = total_spaces // gaps
                    extra_spaces = total_spaces % gaps

                    line = ""
                    for i in range(gaps):
                        line += cur_line[i]
                        # Base spaces add karo
                        line += ' ' * space_per_gap
                        # Agar extra spaces bache hain, to ek ek karke left gaps me baato
                        if i < extra_spaces:
                            line += ' '

                    # Aakhri word loop ke baad add karo (uske baad space nahi chahiye)
                    line += cur_line[-1]
                    res.append(line)

                # --- Reset for next line ---
                cur_line = []
                cur_length = 0

            # Word ko current line me add karo
            cur_line.append(word)
            cur_length += len(word)

        # --- Handle Last Line (Special Case) ---
        # Last line left justified hoti hai (words ke beech 1 space)
        last_line = " ".join(cur_line)
        # Baki bache spaces end me add karo
        remaining_spaces = maxWidth - len(last_line)
        last_line += ' ' * remaining_spaces
        res.append(last_line)

        return res
```

## 69. Sqrt(x)

Given a non-negative integer x, return the square root of x rounded down to the nearest integer. The returned integer should be non-negative as well.

You must not use any built-in exponent function or operator.

For example, do not use pow(x, 0.5) in c++ or x ** 0.5 in python.


Example 1:

Input: x = 4
Output: 2
Explanation: The square root of 4 is 2, so we return 2.
Example 2:

Input: x = 8
Output: 2
Explanation: The square root of 8 is 2.82842..., and since we round it down to the nearest integer, 2 is returned.


Constraints:

- 0 <= x <= 231 - 1

```python
class Solution(object):
    def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        if x < 2:
            return x

        left, right = 2, x // 2

        # Hum 'right' ko return karenge kyunki loop khatam hone ke baad
        # 'right' wo largest integer hoga jiska square <= x hai.
        while left <= right:
            mid = (left + right) // 2
            square = mid * mid

            if square == x:
                return mid
            elif square > x:
                right = mid - 1
            else:
                left = mid + 1

        return right
```

## Newton's Method (Advanced) ⚡
- Yeh method Calculus par based hai aur computer graphics/scientific calculations mein use hota hai. Isse Newton-Raphson Formula kehte hain.Formula:$$root_{new} = \frac{1}{2} \times (root_{old} + \frac{x}{root_{old}})$$Hum ek guess (r) lete hain (usually x), aur is formula ko baar-baar lagate hain jab tak value stabilize na ho jaye.

```python
class Solution(object):
    def mySqrt(self, x):
        if x < 2:
            return x

        r = x
        # Jab tak r*r x se bada hai, r ko shrink karte raho
        while r * r > x:
            r = (r + x // r) // 2

        return r
```

## 70. Climbing Stairs


You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?



Example 1:

Input: n = 2
Output: 2
Explanation: There are two ways to climb to the top.
1. 1 step + 1 step
2. 2 steps
Example 2:

Input: n = 3
Output: 3
Explanation: There are three ways to climb to the top.
1. 1 step + 1 step + 1 step
2. 1 step + 2 steps
3. 2 steps + 1 step


Constraints:

- 1 <= n <= 45

```python
class Solution(object):
    def climbStairs(self, n):
        """
        :type n: int
        :rtype: int
        """
        if n <= 2:
            return n

        # Base cases
        prev2 = 1  # n-2 (Start me step 1 ke ways)
        prev1 = 2  # n-1 (Start me step 2 ke ways)

        for i in range(3, n + 1):
            current = prev1 + prev2

            # Values shift karo aage ke liye
            prev2 = prev1
            prev1 = current

        return prev1
```

```python
class Solution(object):
    def climbStairs(self, n):
        if n <= 2:
            return n

        # Ek array banao n+1 size ka
        dp = [0] * (n + 1)

        # Base Cases set karo
        dp[1] = 1
        dp[2] = 2

        # Loop chalao 3 se n tak
        for i in range(3, n + 1):
            dp[i] = dp[i-1] + dp[i-2]

        return dp[n]
```

```python
class Solution(object):
    def climbStairs(self, n):
        # Ek dictionary/map banao cache ke liye
        memo = {}

        def helper(step):
            # Base Cases
            if step <= 2:
                return step

            # Agar result pehle se calculate kiya hua hai, wahi return karo
            if step in memo:
                return memo[step]

            # Naya result calculate karo aur store karo
            memo[step] = helper(step - 1) + helper(step - 2)
            return memo[step]

        return helper(n)
```

## 71. Simplify Path


You are given an absolute path for a Unix-style file system, which always begins with a slash '/'. Your task is to transform this absolute path into its simplified canonical path.

The rules of a Unix-style file system are as follows:

A single period '.' represents the current directory.
A double period '..' represents the previous/parent directory.
Multiple consecutive slashes such as '//' and '///' are treated as a single slash '/'.
Any sequence of periods that does not match the rules above should be treated as a valid directory or file name. For example, '...' and '....' are valid directory or file names.
The simplified canonical path should follow these rules:

The path must start with a single slash '/'.
Directories within the path must be separated by exactly one slash '/'.
The path must not end with a slash '/', unless it is the root directory.
The path must not have any single or double periods ('.' and '..') used to denote current or parent directories.
Return the simplified canonical path.



Example 1:

Input: path = "/home/"

Output: "/home"

Explanation:

The trailing slash should be removed.

Example 2:

Input: path = "/home//foo/"

Output: "/home/foo"

Explanation:

Multiple consecutive slashes are replaced by a single one.

Example 3:

Input: path = "/home/user/Documents/../Pictures"

Output: "/home/user/Pictures"

Explanation:

A double period ".." refers to the directory up a level (the parent directory).

Example 4:

Input: path = "/../"

Output: "/"

Explanation:

Going one level up from the root directory is not possible.

Example 5:

Input: path = "/.../a/../b/c/../d/./"

Output: "/.../b/d"

Explanation:

"..." is a valid name for a directory in this problem.



Constraints:

- 1 <= path.length <= 3000
- path consists of English letters, digits, period '.', slash '/' or '_'.
- path is a valid absolute Unix path.



```python
class Solution(object):
    def simplifyPath(self, path):
        """
        :type path: str
        :rtype: str
        """
        # Step 1: Path ko '/' se split karo
        # Example: "/a//b/../c" -> ['', 'a', '', 'b', '..', 'c']
        parts = path.split("/")

        stack = []

        # Step 2: Har part par loop chalao
        for part in parts:
            # Case A: Agar part empty hai ("") ya current directory (".") hai
            # Inhe ignore karna hai
            if part == "" or part == ".":
                continue

            # Case B: Agar parent directory ("..") hai
            # To hume piche (up) jana hai, isliye stack se POP karo
            elif part == "..":
                if stack:
                    stack.pop()

            # Case C: Valid directory name hai (jaise "home", "a", "b")
            # To hume andar (down) jana hai, isliye stack me PUSH karo
            else:
                stack.append(part)

        # Step 3: Stack ke words ko wapas '/' se jod do
        # Start me '/' lagana zaroori hai
        return "/" + "/".join(stack)
```

## 72. Edit Distance


Given two strings word1 and word2, return the minimum number of operations required to convert word1 to word2.

You have the following three operations permitted on a word:

Insert a character
Delete a character
Replace a character


Example 1:

Input: word1 = "horse", word2 = "ros"
Output: 3
Explanation:
horse -> rorse (replace 'h' with 'r')
rorse -> rose (remove 'r')
rose -> ros (remove 'e')
Example 2:

Input: word1 = "intention", word2 = "execution"
Output: 5
Explanation:
intention -> inention (remove 't')
inention -> enention (replace 'i' with 'e')
enention -> exention (replace 'n' with 'x')
exention -> exection (replace 'n' with 'c')
exection -> execution (insert 'u')


Constraints:

- 0 <= word1.length, word2.length <= 500
- word1 and word2 consist of lowercase English letters.

```python
class Solution(object):
    def minDistance(self, word1, word2):
        """
        :type word1: str
        :type word2: str
        :rtype: int
        """
        m = len(word1)
        n = len(word2)

        # Step 1: DP Table create karo (m+1 rows, n+1 cols)
        # Dimensions +1 isliye kyunki hum empty string "" ko bhi consider karte hain
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        # Step 2: Base Cases (First Row and First Column)
        # Agar word1 empty hai, to word2 ke saare chars insert karne padenge
        for j in range(n + 1):
            dp[0][j] = j

        # Agar word2 empty hai, to word1 ke saare chars delete karne padenge
        for i in range(m + 1):
            dp[i][0] = i

        # Step 3: Table Fill karo
        for i in range(1, m + 1):
            for j in range(1, n + 1):

                # Case A: Agar characters match karte hain
                # Note: String index 0-based hai, isliye i-1 use kiya
                if word1[i-1] == word2[j-1]:
                    dp[i][j] = dp[i-1][j-1]

                # Case B: Agar match nahi karte -> 1 + Min of 3 operations
                else:
                    insert_op = dp[i][j-1]      # Left
                    delete_op = dp[i-1][j]      # Top
                    replace_op = dp[i-1][j-1]   # Diagonal

                    dp[i][j] = 1 + min(insert_op, delete_op, replace_op)

        # Aakhri cell me final answer hoga
        return dp[m][n]
```

```python
class Solution(object):
    def minDistance(self, word1, word2):
        m, n = len(word1), len(word2)

        # Optimization: Hamesha chhoti string ko columns (n) banao
        # Taaki space O(min(m, n)) lage
        if m < n:
            return self.minDistance(word2, word1)

        # Previous row (Base case: 0 to n)
        prev = range(n + 1)

        for i in range(1, m + 1):
            # Current row create karo
            # curr[0] hamesha i hoga (delete operations)
            curr = [i] + [0] * n

            for j in range(1, n + 1):
                if word1[i-1] == word2[j-1]:
                    curr[j] = prev[j-1]
                else:
                    curr[j] = 1 + min(curr[j-1],    # Insert (Left)
                                      prev[j],      # Delete (Top)
                                      prev[j-1])    # Replace (Diagonal)

            # Current row ab previous ban jayegi agle iteration ke liye
            prev = curr

        return prev[n]
```

## 73. Set Matrix Zeroes

Given an m x n integer matrix matrix, if an element is 0, set its entire row and column to 0's.

You must do it in place.



Example 1:


Input: matrix = [[1,1,1],[1,0,1],[1,1,1]]
Output: [[1,0,1],[0,0,0],[1,0,1]]
Example 2:


Input: matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
Output: [[0,0,0,0],[0,4,5,0],[0,3,1,0]]


Constraints:

- m == matrix.length
- n == matrix[0].length
- 1 <= m, n <= 200
- -2^31 <= matrix[i][j] <= 2^31 - 1


Follow up:

- A straightforward solution using O(mn) space is probably a bad idea.
- A simple improvement uses O(m + n) space, but still not the best solution.
- Could you devise a constant space solution?

```python
class Solution(object):
    def setZeroes(self, matrix):
        """
        :type matrix: List[List[int]]
        :rtype: None Do not return anything, modify matrix in-place.
        """
        m = len(matrix)
        n = len(matrix[0])
        col0 = 1  # Flag for First Column

        # Step 1: Marking Phase (Traverse Forward)
        for i in range(m):
            # Check if 0 is present in the first column
            if matrix[i][0] == 0:
                col0 = 0

            # Check rest of the row
            for j in range(1, n):
                if matrix[i][j] == 0:
                    matrix[i][0] = 0  # Mark Row
                    matrix[0][j] = 0  # Mark Column

        # Step 2: Update Phase (Traverse Backward)
        # Hum ulta (Bottom-Right to Top-Left) isliye chal rahe hain
        # taaki First Row/Col ke flags last tak safe rahein.
        for i in range(m - 1, -1, -1):
            for j in range(n - 1, 0, -1):
                # Agar Row flag YA Column flag 0 hai, to cell ko 0 karo
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0

            # Finally, First Column ko update karo (using col0 variable)
            if col0 == 0:
                matrix[i][0] = 0
```

## 74. Search a 2D Matrix

You are given an m x n integer matrix matrix with the following two properties:

Each row is sorted in non-decreasing order.
The first integer of each row is greater than the last integer of the previous row.
Given an integer target, return true if target is in matrix or false otherwise.

You must write a solution in O(log(m * n)) time complexity.



Example 1:


Input: matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 3
Output: true
Example 2:


Input: matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 13
Output: false


Constraints:

- m == matrix.length
- n == matrix[i].length
- 1 <= m, n <= 100
- -10^4 <= matrix[i][j], target <= 10^4

```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        """
        :type matrix: List[List[int]]
        :type target: int
        :rtype: bool
        """
        if not matrix:
            return False

        m = len(matrix)     # Rows
        n = len(matrix[0])  # Columns

        # Binary Search ki boundaries define karo (0 to m*n - 1)
        low = 0
        high = (m * n) - 1

        while low <= high:
            mid = (low + high) // 2

            # --- Magic Step: 1D Index to 2D Co-ordinates ---
            row_idx = mid // n
            col_idx = mid % n

            mid_val = matrix[row_idx][col_idx]

            if mid_val == target:
                return True
            elif mid_val < target:
                low = mid + 1
            else:
                high = mid - 1

        return False
```

## Virtual 1D Binary Search (Best & Optimal)
- 🏆Time: $O(\log(m \times n))$ | Space: $O(1)$Isme hum matrix ko mentally ek 1D array maante hain aur seedha Binary Search lagate hain.

Formula:

row = mid // cols

col = mid % cols


```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        if not matrix: return False

        rows, cols = len(matrix), len(matrix[0])
        low, high = 0, (rows * cols) - 1

        while low <= high:
            mid = (low + high) // 2
            mid_val = matrix[mid // cols][mid % cols]

            if mid_val == target:
                return True
            elif mid_val < target:
                low = mid + 1
            else:
                high = mid - 1
        return False
```

## Staircase Search (Top-Right Start)
-  🪜Time: $O(m + n)$ | Space: $O(1)$Yeh approach thodi alag hai. Yeh tab use hoti hai jab matrix row-wise aur column-wise sorted ho (LeetCode Problem: "Search a 2D Matrix II").
- Lekin ye yahan bhi kaam karegi.

Logic:Top-Right Corner (0, n-1) par khade ho jao.Agar Target < Current Element: Iska matlab poora column bekar hai (niche sab bade honge).
Left (col--) move karo.Agar Target > Current Element: Iska matlab poori row bekar hai (left mein sab chhote honge). Down (row++) move karo.

Note: Yeh $O(m+n)$ hai, jo $O(\log(mn))$ se slow hai, lekin code bahut simple hai.

```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        if not matrix: return False

        rows, cols = len(matrix), len(matrix[0])

        # Start at Top-Right corner
        row = 0
        col = cols - 1

        while row < rows and col >= 0:
            curr = matrix[row][col]

            if curr == target:
                return True
            elif curr > target:
                col -= 1  # Target chhota hai, left aao
            else:
                row += 1  # Target bada hai, niche jao

        return False
```
