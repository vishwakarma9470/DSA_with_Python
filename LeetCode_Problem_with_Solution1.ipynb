## 26. Remove Duplicates From Sorted Arrays


Given an integer array nums sorted in non-decreasing order, remove the duplicates in-place such that each unique element appears only once. The relative order of the elements should be kept the same.

Consider the number of unique elements in nums to be k​​​​​​​​​​​​​​. After removing duplicates, return the number of unique elements k.

The first k elements of nums should contain the unique numbers in sorted order. The remaining elements beyond index k - 1 can be ignored.

Custom Judge:

The judge will test your solution with the following code:

    assert nums[i] == expectedNums[i];
}
If all assertions pass, then your solution will be accepted.



Example 1:

Input: nums = [1,1,2]
Output: 2, nums = [1,2,_]
Explanation: Your function should return k = 2, with the first two elements of nums being 1 and 2 respectively.
It does not matter what you leave beyond the returned k (hence they are underscores).

Example 2:

Input: nums = [0,0,1,1,1,2,2,3,3,4]
Output: 5, nums = [0,1,2,3,4,_,_,_,_,_]
Explanation: Your function should return k = 5, with the first five elements of nums being 0, 1, 2, 3, and 4 respectively.
It does not matter what you leave beyond the returned k (hence they are underscores).


Constraints:

- 1 <= nums.length <= 3 * 104
- -100 <= nums[i] <= 100
- nums is sorted in non-decreasing order.

```python
class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        # Edge Case: Agar array khali hai (waise constraints me len >= 1 hai)
        if not nums:
            return 0

        # 'l' (left) pointer index 1 se shuru hoga
        # Kyunki index 0 wala number hamesha unique hi maana jayega
        l = 1

        # 'r' (right) pointer loop karega 1 se lekar end tak
        for r in range(1, len(nums)):
            # Agar current number pichle number se alag hai (New Unique Found)
            if nums[r] != nums[r - 1]:
                nums[l] = nums[r] # Unique number ko sahi jagah ('l') par copy karo
                l += 1            # Agli position ke liye 'l' badhao

        # 'l' humein batayega ki total kitne unique elements hain
        return l
```

## 27. Remove Elements


Given an integer array nums and an integer val, remove all occurrences of val in nums in-place. The order of the elements may be changed. Then return the number of elements in nums which are not equal to val.


Consider the number of elements in nums which are not equal to val be k, to get accepted, you need to do the following things:

Example 1:

Output: 2, nums = [2,2,_,_]
Explanation: Your function should return k = 2, with the first two elements of nums being 2.
It does not matter what you leave beyond the returned k (hence they are underscores).
Example 2:

Input: nums = [0,1,2,2,3,0,4,2], val = 2
Output: 5, nums = [0,1,4,0,3,_,_,_]
Explanation: Your function should return k = 5, with the first five elements of nums containing 0, 0, 1, 3, and 4.
Note that the five elements can be returned in any order.
It does not matter what you leave beyond the returned k (hence they are underscores).


Constraints:

- 0 <= nums.length <= 100
- 0 <= nums[i] <= 50
- 0 <= val <= 100


```python
class Solution(object):
    def removeElement(self, nums, val):
        """
        :type nums: List[int]
        :type val: int
        :rtype: int
        """
        k = 0  # Yeh pointer track karega ki valid elements kahan tak hain

        for i in range(len(nums)):
            # Agar current element 'val' nahi hai, toh use rakh lo
            if nums[i] != val:
                nums[k] = nums[i]
                k += 1

        # 'k' ab count ban gaya hai ki kitne elements bache hain
        return k
```

## 28. Find the index of the first occurence in a string

Given two strings needle and haystack, return the index of the first occurrence of needle in haystack, or -1 if needle is not part of haystack.



Example 1:

Input: haystack = "sadbutsad", needle = "sad"
Output: 0
Explanation: "sad" occurs at index 0 and 6.
The first occurrence is at index 0, so we return 0.
Example 2:

Input: haystack = "leetcode", needle = "leeto"
Output: -1
Explanation: "leeto" did not occur in "leetcode", so we return -1.


Constraints:

- 1 <= haystack.length, needle.
- length <= 104
haystack and needle consist of only lowercase English characters.

```python
class Solution(object):
    def strStr(self, haystack, needle):
        """
        :type haystack: str
        :type needle: str
        :rtype: int
        """
        # Agar needle khali hai, toh rule ke hisab se 0 return karte hain
        # (Lekin constraints me length >= 1 hai, toh ye shayad na aaye)
        if not needle:
            return 0

        n = len(haystack)
        m = len(needle)

        # Loop humein poore haystack par nahi chalana,
        # Sirf wahan tak jahan tak needle fit aa sake.
        # Example: haystack="hello", needle="ll" (len 2)
        # Hum sirf index 3 tak check karenge ('l' tak), 'o' ke aage jagah nahi bachi.
        for i in range(n - m + 1):

            # Check karo ki kya current window needle ke barabar hai?
            if haystack[i : i + m] == needle:
                return i

        # Agar loop khatam ho gaya aur kuch nahi mila
        return -1
```

## 29. Divide two Integers


Given two integers dividend and divisor, divide two integers without using multiplication, division, and mod operator.

The integer division should truncate toward zero, which means losing its fractional part. For example, 8.345 would be truncated to 8, and -2.7335 would be truncated to -2.

Return the quotient after dividing dividend by divisor.

Note: Assume we are dealing with an environment that could only store integers within the 32-bit signed integer range: [−231, 231 − 1]. For this problem, if the quotient is strictly greater than 231 - 1, then return 231 - 1, and if the quotient is strictly less than -231, then return -231.



Example 1:

Input: dividend = 10, divisor = 3
Output: 3
Explanation: 10/3 = 3.33333.. which is truncated to 3.
Example 2:

Input: dividend = 7, divisor = -3
Output: -2
Explanation: 7/-3 = -2.33333.. which is truncated to -2.


Constraints:

- -231 <= dividend, divisor <= 231 - 1
- divisor != 0

```python
class Solution(object):
    def divide(self, dividend, divisor):
        """
        :type dividend: int
        :type divisor: int
        :rtype: int
        """
        # 32-bit integer range constants
        MIN_INT = -2**31
        MAX_INT = 2**31 - 1

        # Edge Case: Overflow handle karna (Example: -2147483648 / -1)
        if dividend == MIN_INT and divisor == -1:
            return MAX_INT

        # Result ka sign determine karo
        # Agar dono me se sirf koi ek negative hai, to result negative hoga
        negative = (dividend < 0) != (divisor < 0)

        # Calculation aasan karne ke liye dono ko positive bana lo
        a = abs(dividend)
        b = abs(divisor)
        quotient = 0

        # Jab tak dividend (a) divisor (b) se bada ya barabar hai
        while a >= b:
            temp_divisor = b
            count = 1

            # Divisor ko tab tak double karo (Left Shift << 1)
            # jab tak wo Dividend se chhota hai
            while a >= (temp_divisor << 1):
                temp_divisor <<= 1
                count <<= 1

            # Dividend me se calculation kiya hua bada tukda ghatao
            a -= temp_divisor
            # Quotient me utna count add karo
            quotient += count

        # Sign wapas lagao
        if negative:
            quotient = -quotient

        # Range ke andar return karo (Clamp function)
        return min(max(quotient, MIN_INT), MAX_INT)
```
